# Startup Valuation Is Right

## An Autonomous Agentic Framework for Startup Valuation Opportunities

This notebook adapts the Week 8 "The Price Is Right" product pricing system to **startup valuation estimation**.

**Architecture:**
- **Scanner Agent** — scrapes startup/news RSS feeds, uses structured outputs to parse profiles
- **Specialist Agent** — calls our fine-tuned valuation model deployed on Modal.com
- **Frontier Agent** — RAG pipeline: finds similar startups in ChromaDB, asks GPT to estimate valuation
- **Ensemble Agent** — combines specialist + frontier estimates
- **Messaging Agent** — sends push notifications for high-gap valuation opportunities
- **Planning Agent** — orchestrates everything with an agent loop + tool calling

**Key adaptations from the course:**
- Products -> Startup profiles
- Product prices -> Startup valuations
- Deal discount -> Valuation gap (reported vs estimated)
- DealNews RSS -> TechCrunch / EU-Startups / SaaStr RSS

## Part 3: Populate ChromaDB with Job Data

For the RAG pipeline, we need a vector store of jobs with known valuations.
We load the `lukebarousse/data_jobs` dataset, encode each job with
`all-MiniLM-L6-v2`, and store in ChromaDB.

**This only needs to run once** — the data persists in `startups_vectorstore/`.

```python
import chromadb
from sentence_transformers import SentenceTransformer
from datasets import load_dataset
from tqdm import tqdm

DB_PATH = "startups_vectorstore"

client = chromadb.PersistentClient(path=DB_PATH)
collection = client.get_or_create_collection("startups")
encoder = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

print(f"ChromaDB collection currently has {collection.count()} items")
```

```python
# Set to True to use the full dataset, False for a faster ~20k subset
LITE_MODE = True

dataset = load_dataset("lukebarousse/data_jobs", split="train")
print(f"Loaded {len(dataset)} rows from the dataset")
print(f"Columns: {dataset.column_names}")
```

```python
# Preview a sample row to see the field names
sample = dataset[0]
for key, value in sample.items():
    val_str = str(value)[:100] if value else "None"
    print(f"  {key}: {val_str}")
```

```python
MIN_SALARY = 15_000
MAX_SALARY = 500_000

if collection.count() > 0:
    print(f"ChromaDB already populated with {collection.count()} jobs — skipping")
    print("Delete the startups_vectorstore/ folder and re-run to rebuild")
else:
    # Filter to jobs with valid valuations
    valid_rows = []
    for row in dataset:
        valuation = row.get("valuation_year_avg")
        if valuation is not None and valuation == valuation:  # NaN check
            valuation = float(valuation)
            if MIN_SALARY <= valuation <= MAX_SALARY:
                title = row.get("job_title") or row.get("job_title_short") or ""
                if title:
                    valid_rows.append((row, valuation))

    if LITE_MODE:
        valid_rows = valid_rows[:20_000]

    print(f"Processing {len(valid_rows)} jobs with valid valuations...")

    batch_size = 500
    for i in tqdm(range(0, len(valid_rows), batch_size)):
        batch = valid_rows[i:i + batch_size]
        texts = []
        valuations = []
        ids = []

        for j, (row, valuation) in enumerate(batch):
            title = row.get("job_title") or row.get("job_title_short") or ""
            company = row.get("company_name") or ""
            location = row.get("job_location") or ""
            skills = row.get("job_skills") or ""
            schedule = row.get("job_schedule_type") or ""

            text = f"Job Title: {title}"
            if company:
                text += f"\nCompany: {company}"
            if location:
                text += f"\nLocation: {location}"
            if schedule:
                text += f"\nType: {schedule}"
            if skills:
                text += f"\nSkills: {skills}"

            texts.append(text[:500])
            valuations.append(valuation)
            ids.append(str(i + j))

        vectors = encoder.encode(texts).tolist()
        metadatas = [{"valuation": s} for s in valuations]
        collection.add(
            documents=texts,
            embeddings=vectors,
            ids=ids,
            metadatas=metadatas,
        )

    print(f"\nDone! ChromaDB now has {collection.count()} jobs")
```

### Visualize the vector space (optional)
Use t-SNE to project 384-dimensional job vectors down to 2D.

```python
import numpy as np
from sklearn.manifold import TSNE
import plotly.express as px

sample_size = min(2000, collection.count())
result = collection.get(include=["embeddings", "documents", "metadatas"], limit=sample_size)
vectors = np.array(result["embeddings"])
valuations_viz = [m["valuation"] for m in result["metadatas"]]

# Bin valuations into ranges for coloring
valuation_bins = []
for s in valuations_viz:
    if s < 60_000:
        valuation_bins.append("< $60k")
    elif s < 100_000:
        valuation_bins.append("$60k-$100k")
    elif s < 150_000:
        valuation_bins.append("$100k-$150k")
    elif s < 200_000:
        valuation_bins.append("$150k-$200k")
    else:
        valuation_bins.append("$200k+")

tsne = TSNE(n_components=2, random_state=42, perplexity=30)
reduced = tsne.fit_transform(vectors)

fig = px.scatter(
    x=reduced[:, 0], y=reduced[:, 1], color=valuation_bins,
    title=f"Job Vectors in 2D (t-SNE) — {sample_size} jobs",
    labels={"x": "t-SNE 1", "y": "t-SNE 2", "color": "Salary Range"},
    width=900, height=600,
)
fig.update_traces(marker=dict(size=3))
fig.show()
```

## Part 7: The Agent Loop — Autonomous Planning Agent

This is the heart of the agentic framework. The Planning Agent is an LLM
(GPT-4o) equipped with three tools:
1. `scan_for_job_listings` — calls the Scanner Agent
2. `estimate_market_valuation` — calls the Ensemble Agent
3. `notify_user_of_opportunity` — calls the Messaging Agent

The LLM autonomously decides which tools to call in a `while not done` loop.

### First: fake functions to prove the loop works

```python
from openai import OpenAI
import json

openai_client = OpenAI()

# Test data from the scanner
test_results = scanner.test_scan()
```

```python
# Three FAKE functions — they just print and return dummy data

def scan_for_job_listings() -> str:
    print("[FAKE] Scanning for startup profiles...")
    return test_results.model_dump_json()

def estimate_market_valuation(description: str) -> str:
    print(f"[FAKE] Estimating valuation for: {description[:50]}...")
    return f"The estimated market valuation for this role is $100000"

def notify_user_of_opportunity(description: str, offered_valuation: float, estimated_valuation: float, url: str) -> str:
    print(f"[FAKE] Notifying user: {description[:50]}... offered=${offered_valuation:,.0f}, market=${estimated_valuation:,.0f}")
    return "Notification sent"
```

```python
# Tool definitions — the JSON specs that describe our tools to the LLM

tools = [
    {"type": "function", "function": {
        "name": "scan_for_job_listings",
        "description": "Scans startup profile feeds and returns promising positions with valuation info",
        "parameters": {"type": "object", "properties": {}, "required": [], "additionalProperties": False},
    }},
    {"type": "function", "function": {
        "name": "estimate_market_valuation",
        "description": "Given a job description, estimate the typical market valuation for that role",
        "parameters": {"type": "object", "properties": {
            "description": {"type": "string", "description": "The job description"},
        }, "required": ["description"], "additionalProperties": False},
    }},
    {"type": "function", "function": {
        "name": "notify_user_of_opportunity",
        "description": "Notify the user about the single best valuation opportunity; only call once",
        "parameters": {"type": "object", "properties": {
            "description": {"type": "string", "description": "The job description"},
            "offered_valuation": {"type": "number", "description": "The valuation offered"},
            "estimated_valuation": {"type": "number", "description": "The estimated market valuation"},
            "url": {"type": "string", "description": "The startup profile URL"},
        }, "required": ["description", "offered_valuation", "estimated_valuation", "url"], "additionalProperties": False},
    }},
]
```

```python
# The tool call handler

tool_mapping = {
    "scan_for_job_listings": scan_for_job_listings,
    "estimate_market_valuation": estimate_market_valuation,
    "notify_user_of_opportunity": notify_user_of_opportunity,
}

def handle_tool_call(message):
    results = []
    for tool_call in message.tool_calls:
        name = tool_call.function.name
        args = json.loads(tool_call.function.arguments)
        tool = tool_mapping.get(name)
        result = tool(**args) if tool else ""
        results.append({"role": "tool", "content": result, "tool_call_id": tool_call.id})
    return results
```

```python
# THE AGENT LOOP — this is the core of agentic AI

messages = [
    {"role": "system", "content": "You find promising job opportunities by scanning listings and estimating whether the offered valuation is above estimated valuation baseline, then notify the user of the best one."},
    {"role": "user", "content": "First, scan for startup profiles. Then estimate the market valuation for each. Then pick the single best opportunity where the offered valuation is much higher than the market estimate, and notify the user. Reply OK when done."},
]

done = False
while not done:
    response = openai_client.chat.completions.create(
        model="gpt-4o-mini", messages=messages, tools=tools
    )
    if response.choices[0].finish_reason == "tool_calls":
        message = response.choices[0].message
        results = handle_tool_call(message)
        messages.append(message)
        messages.extend(results)
    else:
        done = True

print(f"\nAgent completed with: {response.choices[0].message.content}")
```

The agent autonomously decided the order and number of tool calls!
Even though these were fake functions, the LLM drove the workflow.

## Architecture Summary

What just happened across a single run:

| Agent | What it does | Model(s) used |
|-------|-------------|---------------|
| **Scanner** | Scrapes RSS feeds, structured outputs | GPT-4o-mini |
| **Preprocessor** | Rewrites job text into structured format | GPT-4o-mini |
| **Specialist** | Fine-tuned startup valuation estimation | Llama 3.2 3B (on Modal) |
| **Frontier** | RAG lookup + GPT estimation | MiniLM encoder + GPT-4o |
| **Ensemble** | Weighted combination (60/40) | — |
| **Messaging** | Crafts alert + push notification | GPT-4o-mini |
| **Planning** | Agent loop with tool calling | GPT-4o |

**Techniques combined:**
- Fine-tuning (training-time) via the Specialist Agent
- RAG (inference-time) via the Frontier Agent
- Structured outputs for parsing unstructured data
- Tool calling for autonomous agent orchestration
- Ensemble model combining multiple estimation approaches
- Deployment to serverless GPU (Modal.com)
